```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/AirBnB_all_chch.csv")
```

```python
june = df[df["year_month"] == "2026-6"]
```

```python
cutoff = june["number_of_reviews"].quantile(0.9)
cutoff
```

183.2 reviews is the top 10% mark, so any listing at or above that is in the most reviewed 10% of Christchurch listings.

```python
top = june[june["number_of_reviews"] >= cutoff]
```

```python
top["neighbourhood"].value_counts().plot(kind="barh")
plt.title("Top 10% most reviewed listings by ward in June 2026")
plt.xlabel("Number of listings")
plt.ylabel("")
plt.gca().invert_yaxis()
plt.show()
```
