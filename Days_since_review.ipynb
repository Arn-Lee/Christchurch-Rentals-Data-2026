```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/data/AirBnB_all_chch.csv")
```

```python
df["last_review"] = pd.to_datetime(df["last_review"])
```

```python
df.groupby("year_month")["last_review"].max()
```

Above shows the reference date for each month, it is the most recent last_review date from Airbnbs in the Christchurch City area. This avoids negative numbers and is a more streamlined system than looking back into another file on the Inside Airbnb website at the last scraped date which just adds unnecessary steps that do not changing the pattern of the data. 

```python
ref_date = df.groupby("year_month")["last_review"].transform("max")
```

```python
df["days_since_review"] = (ref_date - df["last_review"]).dt.days
```

```python
plt.hist(df["days_since_review"], bins=40, edgecolor="black")
plt.title("Days since last review on Christchurch Airbnb listings")
plt.xlabel("Days since last review")
plt.ylabel("Number of listings")
plt.show()
```

This continues to show the same trend as New Zealand data where it is heavily right skewed.
