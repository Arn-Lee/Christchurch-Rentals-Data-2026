```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/AirBnB_all_chch.csv")
```

```python
cutoff = df["number_of_reviews"].quantile(0.9)
cutoff
```

183 reviews is the top 10% mark, so any listing at or above that is in the most reviewed 10% of Christchurch listings.

```python
top = df[df["number_of_reviews"] >= cutoff]
```

```python
top["neighbourhood"].value_counts().plot(kind="barh")
plt.title("Top 10% most reviewed listings by ward")
plt.xlabel("Number of listings")
plt.ylabel("")
plt.gca().invert_yaxis()
plt.show()
```

Central Ward is far ahead, with Banks Peninsula second. Both are the tourist parts of the district, so the most reviewed listings sit where the visitors are.
